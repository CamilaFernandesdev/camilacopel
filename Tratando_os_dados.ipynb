```python
import pandas as pd
path = '/content/drive/MyDrive/COPEL/Gráficos/Dados logs/vaz20 - default 80.log'
caminho = '/drive/MyDrive/COPEL/Gráficos/Dados logs/vaz20 - default 80.log'
```

```python
#leitura do arquivo txt
df = pd.read_fwf(path)
#Retira a primeira linha com informações vazias
df.drop(0, inplace=True)
```

```python

# Tratando a coluna 'Unnamed: 0' separando as informações
new = df['Unnamed: 0'].str.split("-", n = 3, expand = True)
new
```

```python
df['arquivo'] = new[1]
df['usina'] = new[2]
df.drop(columns=["Unnamed: 0"], inplace = True)
```

```python
df.columns
```

```python
#Filtrando dados por usina
df_furnas = df[df['usina']=='FURNAS' ]
df_gbm = df[df['usina']=='GBM' ]
df_sobradinho = df[df['usina'] == 'SOBRADINHO' ]
df_tucurui = df[df['usina'] == 'TUCURUI' ]
df_furnas.dtypes
```

```python
import matplotlib.pyplot as plt

#Plotando um gráfico de linha
fig, ax = plt.subplots()
ax.plot(df_furnas['arquivo'],
        df_furnas['pos_topn_corr'],
        )

# Legendas
ax.set_title('Análise de FURNAS')
ax.set_xlabel('arquivo de vazões')
ax.set_ylabel('Posição')


plt.show()
```

```python
#Plotando um gráfico de linha
fig, ax = plt.subplots()
ax.plot(df_gbm['arquivo'],
        df_gbm['pos_topn_corr'],
        )

# Legendas
ax.set_title('Análise de GBM')
ax.set_xlabel('arquivo de vazões')
ax.set_ylabel('Posição')


plt.show()
```

```python
#Plotando um gráfico de linha
fig, ax = plt.subplots()
ax.plot(df_sobradinho['arquivo'],
        df_sobradinho['pos_topn_corr'],
        )

# Legendas
ax.set_title('Análise de SOBRADINHO')
ax.set_xlabel('arquivo de vazões')
ax.set_ylabel('Posição')


plt.show()
```
